# 2. Missing value audit

```python
missing = pd.DataFrame(index=["Missing", "Percentage"], columns=dataset.columns)
for columnname in dataset.columns:
    missing.loc["Missing", columnname] = dataset[columnname].isnull().sum()
    missing.loc["Percentage", columnname] = dataset[columnname].isnull().sum() / len(dataset) * 100
missing = missing.T
missing
```
